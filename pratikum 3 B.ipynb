```python
#Dika Adi Saputra
#4232501012
#25/09/2026
#pengoprasian hasil aritmatika bilangan bulat
#SOAL 1
nota1 = int(input("Ketik angka bulat pertama: "))
nota2 = int(input("Ketik angka bulat kedua: "))
print("\n=== HASIL KALKULASI ARITMATIKA ===")
print(f"{nota1} + {nota2}  = {nota1 + nota2}")
print(f"{nota1} - {nota2}  = {nota1 - nota2}")
print(f"{nota1} * {nota2}  = {nota1 * nota2}")
if nota2 != 0:
    print(f"{nota1} / {nota2}  = {nota1 / nota2}")
    print(f"{nota1} // {nota2} = {nota1 // nota2} (pembagian bulat)")
    print(f"{nota1} % {nota2}  = {nota1 % nota2} (modulus)")
else:
    print("Operasi pembagian batal dijalankan karena bernilai 0.")
```

```python
#Dika Adi Saputra
#4232501012
#25/09/2026
#program kelayakan mengemudi dan usia
#SOAL 3

usia = int(input("Berapa usia Anda saat ini? "))
input_sim = input("Apakah Anda sudah memiliki SIM? (True/False): ").strip().lower()

punya_sim = input_sim == "true"

dapat_izin = (usia >= 18) and punya_sim

print(f"\nStatus kelayakan mengemudi sendiri: {dapat_izin}")

barang_a = 50000
barang_b = 30000
potongan_harga = 10000

biaya_salah = barang_a + barang_b - potongan_harga / 2


biaya_benar = (barang_a + barang_b - potongan_harga) / 2

print("\n--- Contoh Dampak Penggunaan Tanda Kurung ---")
print(f"Hasil perhitungan tanpa kurung (SALAH) : Rp {biaya_salah:,.0f}")
print(f"Hasil perhitungan memakai kurung (BENAR) : Rp {biaya_benar:,.0f}")
```

```python
#Dika Adi Saputra
#4232501012
#25/09/2026
#perhitungan efisiensi termal pembangkit
#SOAL 4


input_energi = float(input("Inputkan nilai energi masuk (dalam MJ): "))
output_energi = float(input("Inputkan nilai energi keluar (dalam MJ): "))

persentase_efisiensi = (output_energi / input_energi) * 100

print(f"Kinerja efisiensi termal pada unit ini tercatat sebesar {persentase_efisiensi:.2f}%.")
```
