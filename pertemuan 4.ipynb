```python


#DIKA ADI SAPUTRA
#4232501012
#soal 2

# Program Konversi Suhu

#INPUT
celsius = float(input("Masukkan suhu dalam Celsius (°C): "))

# Rumus Fahrenheit = (Celsius * 9/5) + 32
fahrenheit = (celsius * 9 / 5) + 32

# Rumus Kelvin = Celsius + 273.15
kelvin = celsius + 273.15

# FUNGSI OUTPUT
print("\n=== Hasil Konversi Suhu ===")
print(f"Suhu Celsius   : {celsius:.2f} °C")
print(f"Suhu Fahrenheit: {fahrenheit:.2f} °F")
print(f"Suhu Kelvin    : {kelvin:.2f} K")
```
