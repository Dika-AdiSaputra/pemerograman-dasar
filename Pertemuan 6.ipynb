```python
#DIKA ADI SAPUTRA
#4232501012
#SOAL 3

iridiasi = float(input("Masukkan iridiasi matahari (W/m²): "))
soc      = float(input("Masukkan soc baterai (%): "))

daya_panel = iridiasi * 0.18 * 10
print(f"Estimasi daya panel: {daya_panel:.1f} W")

if iridiasi > 600 and soc < 80:
  mode = "CHARGING - iridiasi tinggi, baterai belum penuh, prioritaskan pengisian"
elif iridiasi > 600 and soc >= 80:
  mode = "GRID - EXPORT - baterai sudah penuh, ekspor kelebihan daya ke jaringan"
elif iridiasi > 200 and soc >= 20:
  mode = "NORMAL - iridiasi dan baterai cukup, operasi mandiri"
elif iridiasi <= 200 and soc >= 50:
  mode = "BATTERY - ONLY - iridiasi rendah, andalkan baterai"
else:
  mode = "EMERGENCY - iridiasi rendah dan baterai hampir habis, kurangi beban"

print(f"iridiasi {iridiasi} W/m², soc {soc}% → Mode: {mode}")
```
