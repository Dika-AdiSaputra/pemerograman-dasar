```python
#DIKA ADI SAPUTRA
#4232501012
#SOAL 1

def hitung_belanja(jumlah_buku, harga_per_buku, persen_diskon, uang_budi, jumlah_orang):
    total = jumlah_buku * harga_per_buku
    if isinstance(persen_diskon, int):
        diskon = (persen_diskon / 100) * total
    else:
        diskon = persen_diskon * total
    bayar      = total - diskon
    kembalian  = uang_budi - bayar
    per_orang  = bayar // jumlah_orang
    sisa       = bayar % jumlah_orang
    uang_cukup = uang_budi >= bayar

    print(f"Total       : {int(total)}")
    print(f"Diskon      : {int(diskon)}")
    print(f"Bayar       : {int(bayar)}")
    print(f"Kembalian   : {int(kembalian)}")
    print(f"Per orang   : {int(per_orang)}")
    print(f"Sisa        : {int(sisa)}")
    print(f"Uang cukup? : {uang_cukup}")

hitung_belanja(
    jumlah_buku   =7,
    harga_per_buku=18500,
    persen_diskon =0.1,
    uang_budi     =150000,
    jumlah_orang  =4
)
```

```python
#DIKA ADI SAPUTRA
#4232501012
#SOAL 2

class AnalisisPLTS:
    def __init__(self, jumlah_panel, daya_wp, jam_penyinaran, efisiensi, kebutuhan, tarif, hari_bulan):
        self.jumlah_panel        = jumlah_panel
        self.daya_wp             = daya_wp
        self.jam_penyinaran      = jam_penyinaran
        self.efisiensi           = efisiensi
        self.kebutuhan           = kebutuhan
        self.tarif               = tarif
        self.hari_bulan          = hari_bulan

    def hitung_energi(self):
        energi_harian  = (self.jumlah_panel * self.daya_wp / 1000) * self.jam_penyinaran * self.efisiensi
        energi_bulanan = energi_harian * self.hari_bulan
        nilai_listrik  = energi_bulanan * self.tarif
        kekurangan     = self.kebutuhan - energi_harian
        mencukupi      = energi_harian >= self.kebutuhan

        return {
            "Energi harian (kWh)" : round(energi_harian, 1),
            "Energi bulanan (kWh)": round(energi_bulanan, 1),
            "Nilai listrik (Rp)"  : round(nilai_listrik, 1),
            "Kekurangan (kWh)"    : round(kekurangan, 1),
            "Mencukupi?": mencukupi }
sistem_rumah    = AnalisisPLTS(
    jumlah_panel=12,
    daya_wp    =450,
    jam_penyinaran    =5,
    efisiensi  =0.8,
    kebutuhan  =25,
    tarif      =1444.70,
    hari_bulan =30 )
hasil = sistem_rumah.hitung_energi()
for kunci, nilai in hasil.items():
    print(f"{kunci:<20}: {nilai}")
```

```python
#DIKA ADI SAPUTRA
#4232501012
#SOAL 3

class AnalisisGenerator:
    def __init__(self, daya, jam, konsumsi, harga, kapasitas, arus, hambatan):
        self.daya      = daya
        self.jam       = jam
        self.konsumsi  = konsumsi
        self.harga     = harga
        self.kapasitas = kapasitas
        self.arus      = arus
        self.hambatan  = hambatan

    def hitung_operasional(self):

        energi       = self.daya * self.jam
        solar        = energi * self.konsumsi
        biaya        = solar * self.harga
        tangki_cukup = self.kapasitas // solar
        sisa_solar   = self.kapasitas % solar
        rugi_kabel   = (self.arus ** 2) * self.hambatan
        rugi_tinggi  = (rugi_kabel > 150) and (self.arus >= 20)

        return {
            "Energi": int(energi),
            "Solar" : solar,
            "Biaya" : biaya,
            "Tangki": float(tangki_cukup),
            "Sisa"  : sisa_solar,
            "Rugi"  : rugi_kabel,
            "Tinggi": rugi_tinggi
        }

gen = AnalisisGenerator(
    daya=200,
    jam=8,
    konsumsi=0.25,
    harga=13500,
    kapasitas=1000,
    arus=20,
    hambatan=0.5
)


hasil = gen.hitung_operasional()
print(f"Energi (kWh)     : {hasil['Energi']}")
print(f"Solar (liter)    : {hasil['Solar']}")
print(f"Biaya (Rp)       : {hasil['Biaya']}")
print(f"Tangki cukup     : {hasil['Tangki']} hari")
print(f"Sisa solar (L)   : {hasil['Sisa']}")
print(f"Rugi kabel (W)   : {hasil['Rugi']}")
print(f"Rugi tinggi?     : {hasil['Tinggi']}")
```
